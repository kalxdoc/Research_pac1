```python
import torch.nn as nn
```

```python
input_tensor = torch.tensor([[0.3471, 0.4547, -0.2356]])
linear_layer = nn.Linear(
                         in_features=3,
                         out_features=2
                         )
output_tensor = linear_layer(input_tensor)
print(output_tensor)
```

```python
in_tensor = torch.Tensor([[2, 3, 6, 7, 9, 3, 2, 1,2,4]])
model=nn.Sequential(nn.Linear(10,20),
                    nn.Linear(20,30),
                    nn.Linear(30,10),
                    nn.Linear(10,2))
output_tensor=model(in_tensor)
print(output_tensor)
total=0
for parameter in model.parameters():
  total+=parameter.numel()
print(total)
```

```python
input_tensor = torch.tensor([[2.4]])
input_tensor2 = torch.tensor([[1.0, -6.0, 2.5, -0.3, 1.2, 0.8]])
sigmoid = nn.Sigmoid()
probability = sigmoid(input_tensor)
print(probability)
softmax = nn.Softmax()
probabilities = softmax(input_tensor2)
print(probabilities)
```

```python
input_tensor3 = torch.Tensor([[3, 4, 6, 7, 10, 12, 2, 3, 6, 8, 9]])

# Update network below to perform a multi-class classification with four labels
model = nn.Sequential(
  nn.Linear(11, 20),
  nn.Linear(20, 12),
  nn.Linear(12, 6),
  nn.Linear(6, 4),
  nn.Softmax(dim=-1)
)

output = model(input_tensor3)
print(output)
```

```python
import numpy as np
import torch.nn.functional as F

y = 1
num_classes = 3

# Create the one-hot encoded vector using NumPy
one_hot_numpy = np.array([0, 1, 0])

# Create the one-hot encoded vector using PyTorch
one_hot_pytorch = F.one_hot(torch.tensor(y), num_classes=3)

print("One-hot vector using NumPy:", one_hot_numpy)
print("One-hot vector using PyTorch:", one_hot_pytorch)
```
