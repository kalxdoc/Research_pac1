```python
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset

data = {
    "animal_name": ["sparrow", "eagle", "cat", "dog", "lizard"],
    "hair": [0, 0, 1, 1, 0],
    "feathers": [1, 1, 0, 0, 0],
    "eggs": [1, 1, 0, 0, 1],
    "milk": [0, 0, 1, 1, 0],
    "predator": [0, 1, 1, 0, 1],
    "legs": [2, 2, 4, 4, 4],
    "tail": [1, 1, 1, 1, 1],
    "type": [0, 0, 1, 1, 2],
}

animals=pd.DataFrame(data)
X = animals.iloc[:, 1:-1].to_numpy()
y = animals.iloc[:, -1].to_numpy()
print('X:', X)
print('y:', y )

dataset = TensorDataset(torch.tensor(X),torch.tensor(y))

input_sample, label_sample = dataset[0]
print('Input sample:', input_sample)
print('Label sample:', label_sample)

from torch.utils.data import DataLoader

dataloader = DataLoader(dataset, batch_size=2, shuffle=True)

# Iterate over the dataloader
for batch_inputs, batch_labels in dataloader:
    print('batch_inputs:', batch_inputs)
    print('batch_labels:', batch_labels)

# Input features size is 7, so the first linear layer needs to be (7, 4)
model = nn.Sequential(
    nn.Linear(7, 4),
    nn.ReLU(),
    nn.Linear(4, 1),
)

# Save original weights of the first layer
original_weights = model[0].weight.clone().detach()
print('\n=== Original Weights of the first layer ===')
print(original_weights)

criterion = nn.MSELoss()
optimizer = optim.SGD(
    model.parameters(), lr=0.01, momentum=0.9
)

num_epochs = 100
for epoch in range(num_epochs):
    for data in dataloader:
       feature, target_batch = data
       optimizer.zero_grad()
       # Convert feature batch to float and pass it into the model
       pred = model(feature.float())
       loss = criterion(pred, target_batch.float().unsqueeze(1))
       loss.backward()
       optimizer.step()

# Save new weights of the first layer
new_weights = model[0].weight.clone().detach()
weight_difference = new_weights - original_weights

print('\n=== New Weights of the first layer after training ===')
print(new_weights)
print('\n=== Weight Difference (What changed: New - Original) ===')
print(weight_difference)
```
